**Section 2 — Return Values and Financial Calculations**

Goal: turn standard finance formulas into reusable functions that return numbers.

**Exercise 4 — Simple Interest**

```python
#Define simple_interest(principal, rate, years) that returns principal * rate * years.
#Call it with 1000, 0.05 and 3, and print the result.
#Tip: rate here is a decimal (0.05 means 5%). Be consistent across all your functions so you never mix 5 and 0.05 by mistake.

def simple_interest(principal, rate, years):
  return principal * rate * years

simple_interest(1000, 0.05, 3)
```

**Exercise 5 — Compound Interest**

```python
#Define compound_value(principal, rate, years) that returns the future value: principal * (1 + rate) ** years.
#Print the value of a $2000 investment at 4% over 10 years.
#Tip: In Python, ** is the power operator — do not use ^, which means something completely different

def compound_value(principal, rate, years):
  return principal * (1 + rate) ** years

compound_value(2000, 0.04, 10)
```

**Exercise 6 — Percentage Return**

```python
#Define pct_return(buy_price, sell_price) that returns the return as a percentage: (sell - buy) / buy * 100.
#Test it with a profitable trade and a losing trade.
#Tip: Dividing by buy_price means the function breaks if buy_price is 0 — you will handle that case in a later exercise.

def pct_return(buy_price, sell_price):
  return (sell_price - buy_price) / buy_price * 100

pct_return(100, 110)
```

**Exercise 7 — Convert a Currency**

```python
#Define convert(amount, exchange_rate) that returns amount * exchange_rate.
#Use it to convert 500 EUR into USD, then print the result rounded to 2 decimals using round().
#Tip: round(value, 2) returns a rounded number you can still use in calculations — keep the rounding for display, not for intermediate steps.

def convert(amount, exchange_rate):
  return amount * exchange_rate

value = convert(500, 0.87)
round(value, 2)
```

**Exercise 8 — Average of a List**

```python
#Define average(values) that takes a list of numbers and returns their mean.
#Use a for loop and an accumulator inside the function (do not use the built-in sum()).
#Test it with a list of daily closing prices.
#Tip: Remember that len(values) gives the number of items — divide the accumulated total by it at the end, after the loop, not inside it.

def average(values):
    total = 0
    for v in values:
        total += v
    return total / len(values)

closing_prices = [101.5, 102.3, 99.8, 100.4, 103.0]

print("Average closing price:", average(closing_prices))
```

**Section 3 — Default Values and Multiple Results**

Goal: make functions more flexible with default arguments and return more than one value.

**Exercise 9 — Default Interest Rate**

```python
#Define future_value(principal, years, rate=0.03) with a default rate of 3%.
#Call it once without the rate argument, and once with rate=0.07.
#Tip: Parameters with a default value must always come after the ones without a default, otherwise Python raises a SyntaxError.

def future_value(principal, years, rate=0.03):
    return principal * (1 + rate) ** years

print("Default rate:", future_value(1000, 5))
print("7% rate:", future_value(1000, 5, rate=0.07))
```

**Exercise 10 — Transaction Fee with a Default**

```python
#Define net_amount(amount, fee_rate=0.01) that returns the amount after subtracting the fee.
#Print the net amount for a 10,000 transfer with the default fee, then with a 2.5% fee.
#Tip: You can pass arguments by name when calling: net_amount(10000, fee_rate=0.025). This is clearer than relying on position.

def net_amount(amount, fee_rate=0.01):
    return amount - amount * fee_rate

print("Default fee:", net_amount(10000))
print("2.5% fee:", net_amount(10000, fee_rate=0.025))
```

**Exercise 11 — Return Two Values at Once**

```python
#Define min_max(prices) that returns both the lowest and the highest price of a list.
#Call it and store the two results in two variables on one line: low, high = min_max(prices)
#Tip: A function returns several values by separating them with a comma in the return statement
#— Python packs them into a tuple you can unpack on the receiving side.

def min_max(prices):
    return min(prices), max(prices)

prices = [101.5, 102.3, 99.8, 100.4, 103.0]

low, high = min_max(prices)

print("Low:", low, "| High:", high)
```

**Section 4 — Functions Combining Conditions and Loops**

Goal: reuse if / for / while inside functions, the way real financial tools are built.

**Exercise 12 — Loan Approval Function**

```python
#Define loan_decision(credit_score) that returns the string "Approved" if the score is 650 or above, and "Declined" otherwise.
#Print the decision for three different scores.
#Tip: Use return inside each branch. Once a return runs, the function stops immediately — no need for an else in some cases.

def loan_decision(credit_score):
    if credit_score >= 650:
        return "Approved"
    else: return "Declined"

print(720, loan_decision(720))
print(650, loan_decision(650))
print(580, loan_decision(580))
```

**Exercise 13 — Count Profitable Trades**

```python
#Define count_winners(pnl_list) that takes a list of profit and loss values and returns how many of them are strictly positive.
#Test it with [50, -20, 30, -10, 0, 75].
#Tip: Initialise your counter to 0 before the loop, increment it inside the if, and return it after the loop finishes.

def count_winners(pnl_list):
    count = 0
    for pnl in pnl_list:
        if pnl > 0:
            count += 1
    return count

print("Profitable trades:", count_winners([50, -20, 30, -10, 0, 75]))
```

**Exercise 14 — Years to Reach a Savings Goal**

```python
#Define years_to_goal(principal, rate, goal) that uses a while loop to grow the principal at the given rate until it
#reaches or exceeds the goal.
#The function returns the number of years needed.
#Test it with 1000, 0.06 and 2000.
#Tip: Make sure the balance changes inside the loop, otherwise the condition never becomes False and
#your program runs forever.

def years_to_goal(principal, rate, goal):
    balance = principal
    years = 0
    while balance < goal:
        balance = balance * (1 + rate)
        years += 1
    return years


years_to_goal(1000, 0.06, 2000)
```
