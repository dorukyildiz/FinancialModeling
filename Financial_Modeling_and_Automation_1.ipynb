**Exercise 11 — Profit, Loss, or Breakeven**

```python
#Create a variable pnl (profit and loss), e.g. pnl = -120.5.
#Print "Profit" if pnl > 0, "Loss" if pnl < 0, or "Breakeven" if pnl == 0.
#Tip: Comparing floats to exactly 0 can be risky in real applications due to rounding errors, but it's fine for this exercise.

pnl = float(input("PNL: "))

if pnl > 0:
    print("Profit")
if pnl < 0:
    print("Loss")
if pnl == 0:
    print("Breakeven")
```

**Exercise 17 — Sum of Daily Returns**

```python
#Create a list returns = [0.01, -0.02, 0.015, 0.005, -0.01].
#Use a for loop and an accumulator variable to compute and print the total return

returns = [0.01, -0.02, 0.015, 0.005, -0.01]

total = 0

for r in returns:
    total = total + r

print(total)
```

**Exercise 18 — Running Total of Expenses**


```python
#Create a list expenses = [200, 450, 130, 90].
#Use for i, e in enumerate(expenses) to print each expense with its position, and print the running total after each one

expenses = [200, 450, 130, 90]

running_total = 0

for i, e in enumerate(expenses, start=1):
    running_total += e
    print(i, e, running_total)
```

**Exercise 19 — Count Profitable Days**

```python
#Create a list daily_pnl = [50, -20, 30, -10, 0, 75].
#Use a for loop with an if inside it to count how many days had pnl > 0, and print the count at the end.
#Tip: Combine a counter (count = 0, then count += 1 inside the if) with the loop — this pattern appears constantly when screening financial data.

daily_pnl = [50, -20, 30, -10, 0, 75]

count = 0

for pnl in daily_pnl:
    if pnl > 0:
        count += 1

print(count)
```

**Section 5 — while loops**

Goal: repeat an action until a condition is met, rather than a fixed number of times.

**Exercise 20 — Compound Growth to a Target**

```python
#Start with balance = 1000 and rate = 0.06.
#Use a while loop that multiplies balance by (1 + rate) once per year until balance exceeds 2000.
#Print the number of years it took.

balance = 1000
rate = 0.06
years = 0

while balance <= 2000:
    balance = balance * (1 + rate)
    years += 1

print(years)
```

**Exercise 21 — Loan Amortization Countdown**

```python
#Start with balance = 5000 and payment = 450.
#Use a while loop that subtracts payment from balance each round and prints the remaining balance, stopping once balance <= 0.

balance = 5000
payment= 450


while balance > 0:
    balance = balance - payment
    print("Remaining balance:", balance)

```

**Exercise 22 — Validate a Positive Amount**

```python
#Use a while loop to repeatedly ask the user "Enter a deposit amount: " until they enter a number greater than 0.
#Once valid, print "Accepted: " followed by the amount.
#Tip: Convert inside the loop with float(input(...)) and check the condition each time — this
#input-validation pattern is one of the most common uses of while.

amount = float(input("Enter a deposit amount: "))

while amount <= 0:
    amount = float(input("Enter a deposit amount: "))

print("Accepted:", amount)
```

**Exercise 23 — Savings Goal Tracker**

```python
#Start with savings = 0, monthly_deposit = 200, and goal = 3000.
#Use a while loop to add monthly_deposit to savings each month and print the month number and current savings, stopping once savings >= goal.
#Print the total number of months it took to reach the goal.

savings = 0
monthly_deposit = 200
goal = 3000

month = 0

while savings < goal:
    savings += monthly_deposit
    month += 1
    print("Month", month, "| Savings:", savings)

print("Goal reached in", month, "months")
```
