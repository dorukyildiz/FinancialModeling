**Exercise 3 — Loan Eligibility**

```python
#Ask the user for their monthly income and their monthly debt payments (two input() calls, converted to float).
#Compute the debt ratio: debt / income.
#If the ratio is below 0.35, print "Eligible". Otherwise print "Not eligible".
#Tip: Convert both inputs to float before dividing, or Python will try to divide two strings and raise an error.

monthly_income = float(input("Monthly income: "))

monthly_payment = float(input("Monthly debt payment: "))

ratio = monthly_payment/monthly_income

if (ratio < 0.35):
  print("Eligible")
else:
  print("Not eligible")
```

**Exercise 8 — Validate a Positive Input**

```python
#Use a while loop to repeatedly ask the user "Enter an investment amount: " until they enter a number strictly greater than 0.
#Print "Accepted:" and the amount once it is valid.
#Tip: Convert with float(input(...)) inside the loop, and re-check the condition on every pass.

amount = float(input("Enter an investment amount: "))

while (amount <= 0):
  amount = float(input("Enter an investment amount: "))

  if(amount > 0):
    print("Accepted: ", amount)
```

**Exercise 9 — Monthly Growth Simulation**

```python
#Start with balance = 500 and rate = 0.02.
#Use a while loop that multiplies balance by (1 + rate) once per month until balance reaches or exceeds 1000.
#Print the number of months needed.
#Tip: Make sure balance actually changes inside the loop, otherwise the condition never becomes False.

balance = 500
rate = 1.02

month = 0

while(balance<1000):
  balance *= rate
  month += 1

print("The number of months needed: ", month)
```

**Exercise 11 — Function with a Default Parameter**

```python
#Define a function apply_fee(amount, fee_rate=0.02) that returns amount after subtracting the fee: amount * (1 - fee_rate).
#Call it once without fee_rate, and once with fee_rate=0.05.

def apply_fee(amount, fee_rate=0.02):
  return amount * (1 - fee_rate)

print(apply_fee(100))

print(apply_fee(100, 0.05))
```

**Exercise 12 — Function Returning Multiple Values**

```python
#Define a function stats(values) that returns two values: the average of the list, and its range (max - min).
#You may use the built-in sum(), len(), max() and min() functions here.
#Test it with returns = [0.01, -0.02, 0.03, 0.015], and unpack the two results into separate variables.
#Tip: A function can return several values separated by a comma: return avg, spread. Unpack with avg, spread = stats(returns).

def stats(values):
  avg = sum(values) / len(values)
  spread = max(values) - min(values)
  return avg, spread

returns = [0.01, -0.02, 0.03, 0.015]

avg, spread = stats(returns)

print("Average: ", avg)
print("Range: ", spread)
```

**Exercise 13 — Function Combining a Loop and a Condition**

```python
#Define a function count_above(values, threshold) that returns how many elements of the list are strictly greater than
#threshold.
#Test it with daily_pnl = [50, -20, 30, -10, 0, 75] and threshold = 0.

def count_above(values, threshold):
  days=0
  for i in values:
    if(i > threshold):
      days += 1
  return days


daily_pnl = [50, -20, 30, -10, 0, 75]

count_above(daily_pnl, 0)
```

**Exercise 14 — Build a List of Squared Returns**

```python
#returns = [0.01, -0.02, 0.03, 0.015]
#Using a for loop, build a new list called squared where each element is the square of the corresponding return. Print it.
#Tip: Create an empty list first (squared = []), then use squared.append(r ** 2) inside the loop.


returns = [0.01, -0.02, 0.03, 0.015]

squared = []

for i in returns:
  squared.append(i**2)


print(squared)
```

**Exercise 17 — Create a DataFrame from a Dictionary**

```python
#Import pandas as pd.
#Build a dictionary with the keys Ticker, Sector, Price and Shares, using the data below.
#Convert it into a DataFrame called df with pd.DataFrame(), and print it.

import pandas as pd

data = {
    "Ticker": ["AAPL", "MSFT", "GOOG", "XOM", "CVX"],
    "Sector": ["Tech", "Tech", "Tech", "Energy", "Energy"],
    "Price": [189.3, 320.5, 142.0, 110.2, 158.7],
    "Shares": [50, 20, 30, 40, 15],
}

df = pd.DataFrame(data)

print(df)
```

**Exercise 18 — Select and Filter**

```python
#Using the df from Exercise 17:
#Print only the Price column.
#Print only the rows where Price is above 150.
#Tip: Select a column with df["Price"]. Filter rows with a condition inside brackets: df[df["Price"] > 150].

print(df["Price"])

print(df[df["Price"] > 150])
```

**Exercise 19 — Add a Computed Column**

```python
#Using the same df, add a new column called Value equal to Price multiplied by Shares.
#Print the updated DataFrame.
#Tip: df["Value"] = df["Price"] * df["Shares"] works on the whole column at once — no loop needed.


df["Value"] = df["Price"] * df["Shares"]

print(df)
```

**Exercise 20 — Group and Aggregate**

```python
#Using the df from Exercise 19 (with the Value column), compute the average Value per Sector.
#Print the result.

#Tip: df.groupby("Sector")["Value"].mean() groups the rows by Sector first, then averages Value within each group.

df.groupby("Sector")["Value"].mean()
```

**Exercise 22 — Loop Through a Portfolio**

```python
#portfolio = {"AAPL": 50, "MSFT": 20, "GOOG": 30}
#Use a for loop to print each position as: AAPL: 50 shares
#Then compute and print the total number of shares.
#Tip: for ticker, shares in portfolio.items() gives you the key and the value at the same time.

portfolio = {"AAPL": 50, "MSFT": 20, "GOOG": 30}

total_shares = 0

for ticker, shares in portfolio.items():
  print(f"{ticker}: {shares} shares")
  total_shares += shares

print("Total shares: ", total_shares)
```

**Exercise 23 — Check Before Accessing**

```python
#Using the prices dictionary below, ask the user to enter a ticker with input().
#If the ticker is in the dictionary, print its price. Otherwise print "Ticker not found."
#Tip: if ticker in prices: checks the keys. Accessing a key that doesn't exist raises a KeyError, so check first.
#Tip: .upper() converts the input to capital letters, so "aapl" also works.

prices = {"AAPL": 192.0, "MSFT": 320.5, "GOOG": 142.0, "AMZN": 178.4}

ticker = input("Ticker: ").upper()

if (ticker in prices):
  print(f"{ticker}: {prices[ticker]}")
else:
  print("Ticker not found.")
```
